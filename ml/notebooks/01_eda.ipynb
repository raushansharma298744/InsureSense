# InsureSense - Phase 2: Exploratory Data Analysis

**Dataset**: Vehicle Insurance Cross-Sell Prediction
**Samples**: 50,000
**Target**: Response (binary: 0=No Purchase, 1=Purchase)
**Class Distribution**: 87.7% / 12.3%

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set style
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)
pd.set_option('display.max_rows', 100)

# Load data
df = pd.read_csv('../ml/data/raw/train.csv')
print(f'Shape: {df.shape}')
print(f'Columns: {df.columns.tolist()}')
print(f'\nDtypes:\n{df.dtypes}')
```

```python
# Basic info
print('=== DATASET OVERVIEW ===')
print(f'Total samples: {len(df):,}')
print(f'Total features: {len(df.columns) - 1} (excluding ID)')
print(f'Missing values: {df.isnull().sum().sum()}')
print(f'Duplicate rows: {df.duplicated().sum()}')
print(f'Duplicate IDs: {df["id"].duplicated().sum()}')
```

```python
# Target distribution
target_counts = df['Response'].value_counts()
target_pct = df['Response'].value_counts(normalize=True) * 100

print('=== TARGET DISTRIBUTION ===')
for cls in [0, 1]:
    print(f'Response={cls}: {target_counts[cls]:,} ({target_pct[cls]:.2f}%)')

imbalance_ratio = target_counts[0] / target_counts[1]
print(f'Imbalance ratio: {imbalance_ratio:.2f}:1')
```

```python
# Univariate analysis - Numerical features
num_features = ['Age', 'Annual_Premium', 'Vintage', 'Region_Code', 'Policy_Sales_Channel']

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

for idx, feat in enumerate(num_features):
    ax = axes[idx]
    df[feat].hist(bins=50, ax=ax, alpha=0.7, edgecolor='black')
    ax.set_title(f'{feat} Distribution')
    ax.set_xlabel(feat)
    ax.set_ylabel('Frequency')

# Remove empty subplot
axes[-1].remove()
plt.tight_layout()
plt.savefig('../docs/eda_numerical_distributions.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Numerical statistics
num_features = ['Age', 'Annual_Premium', 'Vintage', 'Region_Code', 'Policy_Sales_Channel']
print(df[num_features].describe().T.to_string())
```

```python
# Univariate analysis - Categorical features
cat_features = ['Gender', 'Driving_License', 'Previously_Insured', 'Vehicle_Age', 'Vehicle_Damage']

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

for idx, feat in enumerate(cat_features):
    ax = axes[idx]
    counts = df[feat].value_counts()
    counts.plot(kind='bar', ax=ax, alpha=0.7, edgecolor='black')
    ax.set_title(f'{feat} Distribution')
    ax.set_xlabel(feat)
    ax.set_ylabel('Count')
    ax.tick_params(axis='x', rotation=45)
    # Add count labels
    for i, v in enumerate(counts.values):
        ax.text(i, v + 500, f'{v:,}', ha='center', va='bottom')

axes[-1].remove()
plt.tight_layout()
plt.savefig('../docs/eda_categorical_distributions.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Categorical value counts
for feat in ['Gender', 'Driving_License', 'Previously_Insured', 'Vehicle_Age', 'Vehicle_Damage']:
    print(f'\n=== {feat} ===')
    counts = df[feat].value_counts()
    pcts = df[feat].value_counts(normalize=True) * 100
    for val in counts.index:
        print(f'  {val}: {counts[val]:,} ({pcts[val]:.2f}%)')
```

```python
# Bivariate analysis - Target vs Numerical features
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

num_features = ['Age', 'Annual_Premium', 'Vintage', 'Region_Code', 'Policy_Sales_Channel']

for idx, feat in enumerate(num_features):
    ax = axes[idx]
    for target_val in [0, 1]:
        subset = df[df['Response'] == target_val][feat]
        label = 'Purchase (1)' if target_val == 1 else 'No Purchase (0)'
        ax.hist(subset, bins=50, alpha=0.5, label=label, density=True, edgecolor='black')
    ax.set_title(f'{feat} by Target')
    ax.set_xlabel(feat)
    ax.set_ylabel('Density')
    ax.legend()

axes[-1].remove()
plt.tight_layout()
plt.savefig('../docs/eda_numerical_by_target.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Statistical comparison by target
num_features = ['Age', 'Annual_Premium', 'Vintage', 'Region_Code', 'Policy_Sales_Channel']

for feat in num_features:
    grp0 = df[df['Response'] == 0][feat]
    grp1 = df[df['Response'] == 1][feat]
    print(f'\n=== {feat} ===')
    print(f'  No Purchase (0): mean={grp0.mean():.2f}, median={grp0.median():.2f}, std={grp0.std():.2f}')
    print(f'  Purchase (1):    mean={grp1.mean():.2f}, median={grp1.median():.2f}, std={grp1.std():.2f}')
    diff_pct = ((grp1.mean() - grp0.mean()) / grp0.mean()) * 100
    print(f'  Difference: {diff_pct:+.2f}%')
```

```python
# Bivariate analysis - Target vs Categorical features
cat_features = ['Gender', 'Driving_License', 'Previously_Insured', 'Vehicle_Age', 'Vehicle_Damage']

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

for idx, feat in enumerate(cat_features):
    ax = axes[idx]
    # Calculate purchase rate by category
    purchase_rate = df.groupby(feat)['Response'].mean().sort_values()
    purchase_rate.plot(kind='barh', ax=ax, alpha=0.7, edgecolor='black', color='steelblue')
    ax.set_title(f'Purchase Rate by {feat}')
    ax.set_xlabel('Purchase Rate')
    ax.set_xlim(0, purchase_rate.max() * 1.3)
    # Add percentage labels
    for i, v in enumerate(purchase_rate.values):
        ax.text(v + 0.005, i, f'{v:.3f}', va='center')

axes[-1].remove()
plt.tight_layout()
plt.savefig('../docs/eda_categorical_by_target.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Purchase rate by category with counts
cat_features = ['Gender', 'Driving_License', 'Previously_Insured', 'Vehicle_Age', 'Vehicle_Damage']

for feat in cat_features:
    print(f'\n=== {feat} ===')
    grp = df.groupby(feat)['Response'].agg(['count', 'sum', 'mean']).rename(columns={'count': 'Total', 'sum': 'Purchases', 'mean': 'Rate'})
    grp['Rate'] = grp['Rate'] * 100
    print(grp.to_string())
    print()
```

```python
# Correlation matrix (numerical only)
num_cols = ['Age', 'Annual_Premium', 'Vintage', 'Region_Code', 'Policy_Sales_Channel', 'Driving_License', 'Previously_Insured', 'Response']
corr_matrix = df[num_cols].corr()

plt.figure(figsize=(12, 10))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.3f', cmap='RdBu_r', center=0,
            square=True, linewidths=0.5, cbar_kws={'shrink': 0.8})
plt.title('Correlation Matrix (Numerical Features)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('../docs/eda_correlation_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Correlation with target
target_corr = df[num_cols].corr()['Response'].drop('Response').sort_values(key=abs, ascending=False)
print('=== Correlation with Target (Response) ===')
for feat, corr in target_corr.items():
    print(f'  {feat:25s}: {corr:+.4f}')
```

```python
# Outlier detection - IQR method
num_features = ['Age', 'Annual_Premium', 'Vintage']

for feat in num_features:
    Q1 = df[feat].quantile(0.25)
    Q3 = df[feat].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[feat] < lower) | (df[feat] > upper)]
    print(f'\n=== {feat} ===')
    print(f'  Q1={Q1:.2f}, Q3={Q3:.2f}, IQR={IQR:.2f}')
    print(f'  Bounds: [{lower:.2f}, {upper:.2f}]')
    print(f'  Outliers: {len(outliers):,} ({len(outliers)/len(df)*100:.2f}%)')
    if len(outliers) > 0:
        print(f'  Range: [{outliers[feat].min():.2f}, {outliers[feat].max():.2f}]')
```

```python
# Box plots for outlier visualization
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

for idx, feat in enumerate(['Age', 'Annual_Premium', 'Vintage']):
    df.boxplot(column=feat, by='Response', ax=axes[idx])
    axes[idx].set_title(f'{feat} by Target')
    axes[idx].set_xlabel('Response')
    axes[idx].set_ylabel(feat)

plt.suptitle('')
plt.tight_layout()
plt.savefig('../docs/eda_boxplots_by_target.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# High cardinality categorical analysis
print('=== Region_Code ===')
print(f'Unique values: {df["Region_Code"].nunique()}')
print(f'Top 10: {df["Region_Code"].value_counts().head(10).to_dict()}')

# Purchase rate by Region_Code (top 20)
region_stats = df.groupby('Region_Code')['Response'].agg(['count', 'mean']).rename(columns={'count': 'Total', 'mean': 'Rate'})
region_stats = region_stats[region_stats['Total'] >= 100].sort_values('Rate', ascending=False)
print(f'\nTop 10 regions by purchase rate (min 100 samples):')
print(region_stats.head(10).to_string())
print(f'\nBottom 10 regions by purchase rate (min 100 samples):')
print(region_stats.tail(10).to_string())
```

```python
# Policy_Sales_Channel analysis
print('=== Policy_Sales_Channel ===')
print(f'Unique values: {df["Policy_Sales_Channel"].nunique()}')
print(f'Top 10: {df["Policy_Sales_Channel"].value_counts().head(10).to_dict()}')

channel_stats = df.groupby('Policy_Sales_Channel')['Response'].agg(['count', 'mean']).rename(columns={'count': 'Total', 'mean': 'Rate'})
channel_stats = channel_stats[channel_stats['Total'] >= 50].sort_values('Rate', ascending=False)
print(f'\nTop 10 channels by purchase rate (min 50 samples):')
print(channel_stats.head(10).to_string())
print(f'\nBottom 10 channels by purchase rate (min 50 samples):')
print(channel_stats.tail(10).to_string())
```

```python
# Segment analysis - Key combinations
print('=== Purchase Rate by Key Segments ===')

# Vehicle_Age x Vehicle_Damage
print('\n--- Vehicle_Age x Vehicle_Damage ---')
seg = df.groupby(['Vehicle_Age', 'Vehicle_Damage'])['Response'].mean().unstack() * 100
print(seg.round(2).to_string())

# Previously_Insured x Vehicle_Damage
print('\n--- Previously_Insured x Vehicle_Damage ---')
seg = df.groupby(['Previously_Insured', 'Vehicle_Damage'])['Response'].mean().unstack() * 100
print(seg.round(2).to_string())

# Gender x Previously_Insured
print('\n--- Gender x Previously_Insured ---')
seg = df.groupby(['Gender', 'Previously_Insured'])['Response'].mean().unstack() * 100
print(seg.round(2).to_string())
```

```python
# Age bins analysis
df['Age_Bin'] = pd.cut(df['Age'], bins=[20, 30, 40, 50, 60, 70, 85], labels=['20-29', '30-39', '40-49', '50-59', '60-69', '70+'])

age_bin_stats = df.groupby('Age_Bin')['Response'].agg(['count', 'mean']).rename(columns={'count': 'Total', 'mean': 'Rate'})
age_bin_stats['Rate'] = age_bin_stats['Rate'] * 100
print('=== Purchase Rate by Age Bin ===')
print(age_bin_stats.round(2).to_string())

# Annual_Premium bins
df['Premium_Bin'] = pd.qcut(df['Annual_Premium'], q=5, labels=['Very Low', 'Low', 'Medium', 'High', 'Very High'])
premium_stats = df.groupby('Premium_Bin')['Response'].agg(['count', 'mean']).rename(columns={'count': 'Total', 'mean': 'Rate'})
premium_stats['Rate'] = premium_stats['Rate'] * 100
print('\n=== Purchase Rate by Premium Quintile ===')
print(premium_stats.round(2).to_string())
```

```python
# Vintage analysis
df['Vintage_Bin'] = pd.qcut(df['Vintage'], q=5, labels=['Very New', 'New', 'Medium', 'Old', 'Very Old'])
vintage_stats = df.groupby('Vintage_Bin')['Response'].agg(['count', 'mean']).rename(columns={'count': 'Total', 'mean': 'Rate'})
vintage_stats['Rate'] = vintage_stats['Rate'] * 100
print('=== Purchase Rate by Vintage Quintile ===')
print(vintage_stats.round(2).to_string())
```
